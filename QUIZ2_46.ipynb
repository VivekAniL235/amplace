```python

import numpy as np

A = np.array([[3.0, 1.0],
              [1.0, 2.0]])

b = np.array([7.0, 4.0])


x = np.array([0.0, 0.0])


r = b - A @ x
p = r.copy()


tol = 1e-6
max_iter = 10


print("Iteration 0: x =", x,
      "Residual norm =", np.linalg.norm(r))

# Conjugate Gradient Method
for k in range(1, max_iter + 1):

    Ap = A @ p


    alpha = np.dot(r, r) / np.dot(p, Ap)


    x_new = x + alpha * p


    r_new = r - alpha * Ap

    # Calculate residual norm
    residual_norm = np.linalg.norm(r_new)

    # iteration details
    print(f"Iteration {k}: x = {x_new}, "
          f"Residual norm = {residual_norm:.10e}")

    # Stop if residual norm is below tolerance
    if residual_norm < tol:
        x = x_new
        break

    # Calculate beta
    beta = np.dot(r_new, r_new) / np.dot(r, r)

    # Update search direction
    p = r_new + beta * p

    # next iteration
    x = x_new
    r = r_new

# Part C: Verification
print("Computed solution:", x)

verification = A @ x - b

print("Ax - b =", verification)
print("Verification norm =", np.linalg.norm(verification))
```
